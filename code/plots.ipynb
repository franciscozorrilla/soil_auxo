```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats

# Load your dataset (skip the header row)
data = np.genfromtxt('data/figure_data/figure3panelB.tsv', delimiter='\t', skip_header=1)

# Data: Swap X and Y axes (Predicted on x, Observed on y)
x = data[:, 2]  # Total predicted auxotrophies
y = data[:, 1]  # Total observed auxotrophies

# Sort the data for plotting (optional, for better visualization)
sorted_indices = np.argsort(x)
x_sorted = x[sorted_indices]
y_sorted = y[sorted_indices]

# Fit a linear regression and calculate estimated values
a, b = np.polyfit(x_sorted, y_sorted, 1)
x_extended = np.linspace(min(x_sorted)-0.3, max(x_sorted)+0.5, 100)
y_est_extended = a * x_extended + b

# Calculate the standard error of the estimate
s_err = np.sqrt(np.sum((y_sorted - (a * x_sorted + b)) ** 2) / (len(x_sorted) - 2))

# T-value for 95% confidence interval
t_val = stats.t.ppf(0.975, len(x_sorted) - 2)

# Margin of error for confidence interval
ci = t_val * s_err * np.sqrt(1/len(x_sorted) + (x_extended - np.mean(x_sorted)) ** 2 / np.sum((x_sorted - np.mean(x_sorted)) ** 2))

# Set up the figure size and layout
plt.rcParams["figure.figsize"] = [6.0, 6.0]
plt.rcParams["figure.autolayout"] = True

# Create figure and axis
fig, ax = plt.subplots()

# Plot the linear fit
ax.plot(x_extended, y_est_extended, '-', color='Black')

# Plot the confidence interval (CI) as a gray area around the regression line
ax.fill_between(x_extended, y_est_extended - ci, y_est_extended + ci, color='gray', alpha=0.2)

# Set marker points (open markers with edges)
ax.plot(x_sorted, y_sorted, 'o', markerfacecolor='#DAC883', markeredgecolor='Black', markersize=8)

# Customize fonts
font2 = {'family':'DejaVu Sans', 'color':'black', 'size':16}
plt.xlabel("Total observed auxotrophies", fontdict=font2)
plt.ylabel("Total predicted auxotrophies", fontdict=font2)

# Customize the tick size and thickness
tick_thickness = 2
plt.xticks([0, 5, 10, 15, 20], fontsize=16)  # Show only these values on the x-axis
plt.yticks([0, 5, 10, 15], fontsize=16)  # Correct y-axis ticks including 15
ax.tick_params(axis='both', which='major', width=tick_thickness, direction='out', length=7)

# Set the x and y axis limits
ax.set_xlim([-0.5, 20.5])  # Set the x-axis limits up to 20
ax.set_ylim([-0.3, 15.5])  # Set the y-axis limits, including space below 0

# Hide top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Control x-axis and y-axis spines to not extend beyond the last tick
ax.spines['bottom'].set_bounds(-0.5, 20)
ax.spines['left'].set_bounds(-0.3, 15)

# Set rho value and p-value
rho_value = - 0.071
p_value = 5.9e-01


# Add the text for Spearman correlation coefficient (ρ) and p-value in desired format
plt.text(0.05, 14.0, r'$\rho$ = {:.2f}'.format(rho_value), fontsize=14, ha='left', va='center')
plt.text(0.05, 12.5, r'$P = 5.9 \times 10^{-1}$', fontsize=14, ha='left', va='center')

# Tight layout to ensure no overlap
plt.tight_layout()

# Save the figure as a vector-quality PDF with a transparent background
plt.savefig('outputs/Figure3B.pdf', format='pdf', dpi=300, bbox_inches='tight', transparent=True)

# Display the plot
plt.show()
```
